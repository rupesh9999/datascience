```python
# check unique values in the department column
df['department'].unique()
```

```python
# step1: Handle missing values in 'age' column by filling with mean
print("STEP 1: Handling numerical missing values")
print("\nRecords with missing age before filling:")
print(df[df['age'].isnull()][['age', 'salary', 'department']])
```

```python
print("\Records with missing age after filling:")
print(df[df['salary'].insull()][['age', 'salary', 'department']])
```
