```python
# Multiple plots
## sample data

x = [1,2,3,4,5]
y1 = [1,4,9,16,25]
y2 = [1,2,3,4,5]

plt.figure(figsize = (9,5))
plt.plot(x,y1,color='green')

plt.subplot(2,2,1)
plt.plot(x,y1,color='green')
plt.title("Plot 1")

plt.subplot(2,2,2)
plt.plot(y1,y1,color='green')
plt.title("Plot 2")
```

```python
### Bar plot

categories = ['A','B','C','D','E']
values = [5,7,3,8,6]

## Create a bar plot
plt.bar(categories,values,color='magenta')
plt.xlabel('categories')
plt.ylabel('Values')
plt.title('Bar Plot')
plt.show()
```

__Histograms__

Histograms are used to represent the distribution of a dataset. They divide the data into bins and count the number of data points in each bin.

```python
# sample data

data = [1,2,3,3,4,5,6,7,8,9,9,9,9,9]

## create a histogram
plt.hist(data,bins=5,color='yellow',edgecolor='black')
```

```python
## creating a scatter plot

# sample data

x = [1,2,3,4,5]
y = [2,3,4,5,6]

plt.scatter(x,y,color="blue",marker='x')
```

```python
## pie chart

labels=['A','B','C','D']
sizes=[30,20,40,10]
colors=['gold','yellowgreen','lightcoral','lightskyblue']
explode=(0.2,0,0,0) # move out the 1st slice

## create a pie chart
plt.pie(sizes,explode=explode,labels=labels,colors=colors,autopct="%1.1f%%",shadow=True)
```
