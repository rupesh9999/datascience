__Data Manipulation and Analysis with Pandas__

Data manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform, extract insights from data. In this lesson, we will cover various data manipulation and analysis techniques using pandas.

```python
import pandas as pd
```

```python
df=pd.read_csv('data.csv')
## fetch the first five rows
df.head(5)
```

```python
df.tail(5)
```

```python
df.describe()
```

```python
df.dtypes
```

```python
## Handling missing values
df.isnull().any()
```

```python
df.isnull().any(axis=1)
```

```python
df.isnull().sum()
```

```python
df_filled=df.fillna(0)
```

```python
## filling missing values with the mean of the column
df['Sales_fillNA']=df['Sales'].fillna(df['Sales'].mean())
df
```

```python
## Renaming columns
df=df.rename(columns={'Date': 'Sale Date'})
df.head()
```

```python
## change datatypes
df['Value_new']=df['Value'].astype(float)
df.head()
```

```python
df['New value']=df['Value'].apply(lambda x:x*2)
df.head()
```

```python
## Data Aggregating and grouping
grouped_mean=df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['Product', 'Region'])['Value'].sum()
print(grouped_sum)
```

```python
df.groupby(['Product', 'Region'])['Value'].sum()
```

```python
## aggregate multiple functions
grouped_agg=df.groupby('Region')['Value'].agg(['mean','sum','count'])
grouped_agg
```

```python
## merging and joining Dataframes
# created sample dataframes

df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1,2,3]})
df2 = pd.DataFrame({'Key': ['A','B','D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
## merge Dataframe on the key columns
pd.merge(df1, df2, on="Key", how="inner")
```

```python
pd.merge(df1, df2, on="Key", how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```
